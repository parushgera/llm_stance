```python
import os
import re
import pandas as pd

def create_training_summary_df(base_dir='lora'):
    """
    Parses training files to create a wide-format summary DataFrame with a
    dedicated column for dataset size.

    Args:
        base_dir (str): The root directory containing model folders.

    Returns:
        pandas.DataFrame: A DataFrame with targets as rows. The first column is
                          'Dataset Size', and subsequent columns are model names
                          with their training times as values.
    """
    # --- 1. Define the directory structure ---
    models = [d for d in os.listdir(base_dir) if os.path.isdir(os.path.join(base_dir, d)) and d != 'logs']
    targets = ['at', 'bernie', 'cc', 'dt', 'dtp', 'ent', 'face', 'fauci', 'fm', 'hc', 'hlt', 'joe', 'la', 'school', 'stay']

    # --- 2. Initialize data structures ---
    # Nested dictionary for training times: {'model': {'target': time}}
    training_times = {}
    # Simple dictionary for dataset sizes: {'target': size}
    dataset_sizes = {}

    print("Starting data extraction...")
    # --- 3. Loop through each model and target to extract data ---
    for model_name in sorted(models):
        # Initialize the inner dictionary for the current model
        training_times[model_name] = {}
        for target_name in targets:
            file_path = os.path.join(base_dir, model_name, target_name, 'training_time.txt')

            try:
                with open(file_path, 'r') as f:
                    for line in f:
                        # Extract training time (in minutes)
                        if 'Total training time' in line:
                            match_time = re.search(r'\((\d+\.?\d*)\s+minutes\)', line)
                            if match_time:
                                training_times[model_name][target_name] = float(match_time.group(1))

                        # Extract dataset size
                        elif 'Dataset size' in line:
                            # We only need to store this once per target, but overwriting is harmless
                            match_size = re.search(r'Dataset size:\s*(\d+)', line)
                            if match_size:
                                dataset_sizes[target_name] = int(match_size.group(1))

            except FileNotFoundError:
                print(f"Warning: File not found for model '{model_name}', target '{target_name}'.")
                # Add None to keep the DataFrame structure consistent
                training_times[model_name][target_name] = None
            except Exception as e:
                print(f"Error reading file for model '{model_name}', target '{target_name}': {e}")
                training_times[model_name][target_name] = None

    print("Data extraction complete.")

    # --- 4. Assemble the final DataFrame ---
    # Create the main DataFrame with times (Targets as index, Models as columns)
    df = pd.DataFrame(training_times)
    df.index.name = 'Target'
    
    # Create a Pandas Series from the sizes dictionary
    size_series = pd.Series(dataset_sizes, name='Dataset Size')
    
    # Insert the 'Dataset Size' column at the beginning of the DataFrame
    df.insert(0, 'Dataset Size', size_series)

    # Convert size column to integer, handling potential missing values
    df['Dataset Size'] = df['Dataset Size'].astype('Int64')

    return df

# --- Main execution block ---
if __name__ == "__main__":
    # Create the DataFrame by calling the function
    training_summary_df = create_training_summary_df(base_dir="/storage3-ciber/parush/lora")

    print("\n--- Training Summary ---")
    print(training_summary_df)
```
